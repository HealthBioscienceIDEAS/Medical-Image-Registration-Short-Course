## 4. Normalized Cross Correlation (NCC) and Normalized Mutual Information (NMI)

```python
# Define a function to calculate the NCC between two images
def calcNCC(A, B):
    """
    function to calculate the normalised cross correlation between two images

    SYNTAX:
    NCC = calcNCC(A, B)

    INPUTS:
    A - an image stored as a 2D array
    B - an image stored as a 2D array. B must the the same size as A

    OUTPUTS:
    NCC - the value of the normalised cross correlation

    NOTE
    if either of the images contain NaN values these pixels should be
    ignored when calculating the NCC.
    """

    # TODO: Remove pixels that contain NaN in either image
    nan_inds = np.logical_or(np.isnan(A), np.isnan(B))
    A = A[np.logical_not(nan_inds)]
    B = B[np.logical_not(nan_inds)]

    # TODO: Calculate the mean and std of each image
    mu_A = np.mean(A)
    mu_B = np.mean(B)
    sig_A = np.std(A)
    sig_B = np.std(B)

    # TODO: Calculate and return the NCC
    NCC = np.sum((A - mu_A) * (B - mu_B)) / (A.size * sig_A * sig_B)
    return NCC
```

```python
# do the same as above but this time calculate the NCC, joint and marginal entropies between the original and transformed images
NCCs = np.zeros((theta.size, 4))
H_ABs = np.zeros((theta.size, 4))
H_As = np.zeros((theta.size, 4))
H_Bs = np.zeros((theta.size, 4))

for n in range(theta.size):  
    #TODO: copy your code from the previous cell to create the affine matrix and corresponding deformation field, and to resample the images.
    # create affine matrix and corresponding deformation field
    aff_mat = ut.affineMatrixForRotationAboutPoint(theta[n], [10, 10])
    def_field = ut.defFieldFromAffineMatrix(aff_mat, num_pix_x, num_pix_y)

    # resample the images
    ct_img_int8_resamp = ut.resampImageWithDefField(ct_img_int8, def_field)
    ct_img_int16_resamp = ut.resampImageWithDefField(ct_img_int16, def_field)
    mr_img_int16_resamp = ut.resampImageWithDefField(mr_img_int16, def_field)

    # Calculate the NCC values
    NCCs[n, 0] = calcNCC(ct_img_int16, ct_img_int16_resamp)
    NCCs[n, 1] = calcNCC(ct_img_int16, ct_img_int8_resamp)
    NCCs[n, 2] = calcNCC(ct_img_int16, mr_img_int16_resamp)
    NCCs[n, 3] = calcNCC(ct_img_int8, ct_img_int8_resamp)
    
    # Calculate the joint and marginal entropies
    H_ABs[n, 0], H_As[n, 0], H_Bs[n, 0]  = ut.calcEntropies(ct_img_int16, ct_img_int16_resamp)
    H_ABs[n, 1], H_As[n, 1], H_Bs[n, 1]  = ut.calcEntropies(ct_img_int16, ct_img_int8_resamp)
    H_ABs[n, 2], H_As[n, 2], H_Bs[n, 2]  = ut.calcEntropies(ct_img_int16, mr_img_int16_resamp)
    H_ABs[n, 3], H_As[n, 3], H_Bs[n, 3]  = ut.calcEntropies(ct_img_int8, ct_img_int8_resamp)
  
# Calculate the mutual information and normalised mutual information
MIs = H_As + H_Bs - H_ABs
NMIs = (H_As + H_Bs) / H_ABs
```
